## Step 7 · Fetch live data · 14 min

Here is the idea worth taking away from today: **a dashboard does not contain data. It contains instructions
for going and getting data.** Those instructions are called a **data loader**.

We will use [Open-Meteo](https://open-meteo.com/), a free forecasting service needing no account and no API
key. Make the folder and the file:

````python
mkdir src/data
touch src/data/doha.json.js
````

*(Windows PowerShell: `New-Item src/data/doha.json.js`)*

The double extension is the whole trick: **`.json`** is what this file *produces*, **`.js`** is the language
it is *written in*.

Paste this into `src/data/doha.json.js`:

````python
// EDIT ME — these two numbers decide whose weather this is.
const latitude = 25.2854;   // Doha
const longitude = 51.5310;

const url = "https://api.open-meteo.com/v1/forecast"
  + `?latitude=${latitude}&longitude=${longitude}`
  + "&hourly=temperature_2m,apparent_temperature"
  + "&timezone=auto&forecast_days=7";

const response = await fetch(url);
if (!response.ok) throw new Error(`fetch failed: ${response.status}`);
const data = await response.json();

// The API hands us COLUMNS. A chart needs ROWS. This loop is the reshape.
const rows = data.hourly.time.map((t, i) => ({
  time: t,                          // "2026-10-28T14:00"
  day: t.slice(5, 10),              // "10-28"
  hour: Number(t.slice(11, 13)),    // 14
  temperature: data.hourly.temperature_2m[i],
  feelsLike: data.hourly.apparent_temperature[i]
}));

process.stdout.write(JSON.stringify(rows));
````

**Read it before you run it.** Four things happen:

1. **Two numbers** name a point on Earth — 25.2854°N, 51.5310°E is Doha.
2. `fetch(url)` asks Open-Meteo for seven days of hourly forecast at that point.
3. **The reshape.** The API replies with *columns* — one long list of times, another of temperatures. Charts
   need *rows* — one object per hour with that hour's values together. That `.map()` is the translation, and
   it is the single most common operation in real data work.
4. `process.stdout.write` prints the result, which is how a loader hands back its output.

Run it by hand to see what it makes:

````python
node src/data/doha.json.js
````

A wall of JSON. Ugly, but it is real Doha weather that did not exist yesterday.

## Step 8 · Three charts, three questions · 16 min

We use **Observable Plot**, a charting library built in — no import needed. Each chart is a small function
that takes the data and returns a chart.

In `src/doha.md`, **replace the `display(forecast);` block** with this. It is long, but it is three
independent charts — read one at a time.

````python
```js
// QUESTION 1 — What is this week going to do?
function temperatureLine(data, {width} = {}) {
  return Plot.plot({
    title: "Hourly temperature",
    width,
    height: 260,
    x: {label: null},
    y: {grid: true, inset: 10, label: "°C"},
    marks: [
      Plot.lineY(data, {
        x: d => new Date(d.time),
        y: "temperature",
        stroke: "temperature",
        z: null,
        curve: "step-after"
      })
    ]
  });
}

// QUESTION 2 — When in the day does the heat arrive?
function temperatureHeatmap(data, {width} = {}) {
  return Plot.plot({
    title: "When is it hottest?",
    width,
    height: 420,
    color: {scheme: "YlOrRd", legend: true, label: "°C"},
    x: {label: "Date"},
    y: {label: "Hour of day"},
    marks: [
      Plot.cell(data, {x: "day", y: "hour", fill: "temperature", tip: true})
    ]
  });
}

// QUESTION 3 — Does the thermometer tell the whole story?
function feelsLikeGap(data, {width} = {}) {
  return Plot.plot({
    title: "What the thermometer says vs. what your body feels",
    width,
    height: 420,
    x: {label: null},
    y: {grid: true, label: "°C"},
    marks: [
      Plot.areaY(data, {
        x: d => new Date(d.time),
        y1: "temperature",
        y2: "feelsLike",
        fill: "#f97316",
        fillOpacity: 0.25
      }),
      Plot.lineY(data, {x: d => new Date(d.time), y: "temperature", stroke: "#0ea5e9", strokeWidth: 2}),
      Plot.lineY(data, {x: d => new Date(d.time), y: "feelsLike", stroke: "#f97316", strokeWidth: 2})
    ]
  });
}
```

```js
display(temperatureLine(forecast));
display(temperatureHeatmap(forecast));
display(feelsLikeGap(forecast));
```
````

Save. **Three charts of the same 168 rows.**

**Chart 1 — the line.** A week of temperature. You can see the daily rhythm, but seven days of wiggles is a
lot to read at once.

**Chart 2 — the heatmap.** `Plot.cell` draws one tile per hour: date across, hour of day down, colour = heat.
Same data, but now the *pattern* is unmissable — a dark band in the afternoons, a cool stripe before dawn.
Hover any tile for the exact value. **The line chart cannot show you this, and the heatmap cannot show you
the line's precise shape.** Neither is better. They answer different questions.

**Chart 3 — the gap.** Blue is what the thermometer reads; orange is what it *feels like* once humidity is
accounted for. The shaded band is the difference.

> ### Look hard at chart 3
> The blue line is what an instrument measures. The orange line is what a human body actually experiences.
> The shaded band is the disagreement between them.
>
> **Find the widest part of the band on your own chart.** In late October it is usually a humid afternoon and
> the gap runs three to five degrees. Notice it also goes the *other* way on dry evenings, when the orange
> line dips below the blue one. (Run this same chart in August and the band is twice as wide.)
>
> So: **which number is "the temperature"?** Both are real, both are measured, and they disagree. A chart
> showing only the blue line is not lying — but someone deciding whether outdoor work is safe this afternoon
> would be badly served by it.
>
> **Choosing which of two true numbers to show is a storytelling decision, not a technical one.** No tool and
> no AI makes it for you. That choice is what this entire course is about.

## Step 9 · Lay it out like a dashboard · 5 min

The charts are stacked down the page. Observable gives you a `grid` of `card`s to arrange them.

In `src/doha.md`, **delete the three `display(...)` lines** — otherwise every chart renders twice — and put
this at the bottom of the file instead:

````python
<div class="grid grid-cols-1">
  <div class="card">${resize((width) => temperatureLine(forecast, {width}))}</div>
</div>

<div class="grid grid-cols-2">
  <div class="card">${resize((width) => temperatureHeatmap(forecast, {width}))}</div>
  <div class="card">${resize((width) => feelsLikeGap(forecast, {width}))}</div>
</div>
````

`resize` is why each chart function takes a `width`: it redraws the chart whenever its card changes size.
Drag your browser window narrower and watch the two-column grid fold into one.

> **Why two grids and not one?** Every card in a single grid is forced to the same height. Put the tall
> heatmap beside the short line chart in one grid and the line chart card stretches to match, leaving a
> lake of empty space. Two grids let each row size itself.

Last touch — add this at the **very top** of `src/doha.md`, above the `#` heading, as the first thing in the
file:

````python
---
theme: dashboard
toc: false
---
````

Save. The page goes full-width, and picks up a light or dark palette to match your laptop's system
setting. **That is a dashboard.**
